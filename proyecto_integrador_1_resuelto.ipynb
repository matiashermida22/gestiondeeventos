## 1 - Importo las librerías a utilizar

```python
from datetime import datetime, timezone
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

# 2 - Extraigo la información de la página Whale Alert

```python
url_whale_alert = "https://whale-alert.io/whales.html"
def get_whale_alert_df(url=url_whale_alert):
    response = requests.get(url)
    response.raise_for_status()

    soup = BeautifulSoup(response.content, "html.parser")
    rows = soup.find("table", class_="table").find("tbody").find_all("tr")

    records = []
    for row in rows:
        th = row.find("th", {"scope": "row"})
        img = th.find("img")
        coin = img["alt"].strip() if img else th.get_text(strip=True)

        tds = row.find_all("td")
        if len(tds) < 2:
            continue

        records.append({
            "crypto": coin,
            "known": tds[0].get_text(strip=True),
            "unknown": tds[1].get_text(strip=True),
        })

    return pd.DataFrame(records)
```

```python
whale_alert_raw = get_whale_alert_df()
```
